**01 – Schema Exploration: Bluesky Blocking Data**

Understand the structure, meaning, and quality of each Parquet file before analysis. This notebook focuses on:

- Column names & data types
- Primary keys & foreign keys
- Timestamps & user identifiers
- Missing values & duplicates
- Early hypotheses about relationships between tables

```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
```

```python
blocks = pd.read_parquet("blocks.parquet")
list_blocks = pd.read_parquet("list_blocks.parquet")
lists = pd.read_parquet("lists.parquet")
list_items = pd.read_parquet("list_items.parquet")


posts = pd.read_parquet("chunk_0_posts.parquet")
likes = pd.read_parquet("chunk_0_likes.parquet")
reposts = pd.read_parquet("chunk_0_reposts.parquet")
follows = pd.read_parquet("chunk_0_follow.parquet")
```

```python
def schema_overview(df, name):
    print(f"\n=== {name} ===")
    display(df.head())
    display(df.dtypes)
    print("Rows:", len(df))
    display(df.isna().mean().sort_values(ascending=False))
```

```python
schema_overview(blocks, "blocks")
```

```python
schema_overview(list_blocks, "list_blocks")
```

```python
schema_overview(lists, "lists")
```

```python
schema_overview(list_items, "list_items")
```

```python
schema_overview(posts, "posts")
```

```python
schema_overview(likes, "likes")
```

```python
schema_overview(follows, "follows")
```
