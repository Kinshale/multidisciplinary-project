```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report


MAX_EVENTS = 30
EVENT_DIM = len(dataset_X[0][0])

def pad_and_flatten(sequences, max_events, event_dim):
    X_out = []

    for seq in sequences:
        
        padded_seq = seq + [[0.0] * event_dim] * (max_events - len(seq))

        X_out.append(np.array(padded_seq).flatten())

    return np.array(X_out)


x = pad_and_flatten(dataset_X, MAX_EVENTS, EVENT_DIM)
y = np.array(dataset_Y)

X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)


model = RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))


from joblib import dump

dump(model, "rf_complete.joblib")
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score

def grouped_importance(model, X, y, groups):
    # Ensure we are working with a numpy array for speed/indexing
    X_base = X.values if hasattr(X, 'values') else X.copy()
    
    baseline_score = accuracy_score(y, model.predict(X_base))
    scores = {}
    
    for name, indices in groups.items():
        X_permuted = X_base.copy()
        
        # CORRECT WAY: Shuffle the data within each column
        for idx in indices:
            # Generate a random permutation of indices for the rows
            shuffled_indices = np.random.permutation(len(X_permuted))
            # Reassign the column with the shuffled values
            X_permuted[:, idx] = X_permuted[shuffled_indices, idx]
            
        permuted_score = accuracy_score(y, model.predict(X_permuted))
        scores[name] = baseline_score - permuted_score
        print(f"Done with {name}")
        
    return scores

# Define your groups (example for all 30 events combined)
event_size = 102
groups = {
    "Event Types": [i for i in range(3060) if i % event_size < 22],
    "Dates":       [i for i in range(3060) if i % event_size == 22],
    "Collections": [i for i in range(3060) if 23 <= i % event_size <= 39],
    "Labels":      [i for i in range(3060) if 40 <= i % event_size <= 56],
    "Purpose":     [i for i in range(3060) if 57 <= i % event_size <= 60],
    "Languages":   [i for i in range(3060) if 61 <= i % event_size <= 101],
}

# This will run in seconds, not hours!
importances = grouped_importance(model, X_test, y_test, groups)

df_imp = pd.DataFrame(list(importances.items()), columns=['Category', 'Importance'])
df_imp = df_imp.sort_values(by='Importance', ascending=True)

# Plotting
plt.figure(figsize=(10, 6))
bars = plt.barh(df_imp['Category'], df_imp['Importance'], color='#3498db')

# Add the values at the end of the bars for clarity
for bar in bars:
    width = bar.get_width()
    plt.text(width, bar.get_y() + bar.get_height()/2, 
             f' {width:.5f}', va='center', fontweight='bold')

plt.xlabel('Drop in Accuracy (Permutation Importance)')
plt.title('Feature Importance by Grouped Categories')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()

plt.show()
```
